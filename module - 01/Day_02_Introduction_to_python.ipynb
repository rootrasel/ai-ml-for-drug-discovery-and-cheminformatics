## 3. Python Data Types

A **data type** describes the kind of value stored in a variable. Choosing the correct data type helps Python work with the value appropriately.

| Data type | Meaning | Example |
|---|---|---|
| `str` | Text (string) | `"Aspirin"` |
| `int` | Whole number | `25` |
| `float` | Decimal number | `180.16` |
| `bool` | True or False | `True` |
| `list` | Ordered, changeable collection | `[1, 2, 3]` |
| `tuple` | Ordered, unchangeable collection | `(1, 2, 3)` |
| `dict` | Key-value collection | `{"name": "Aspirin"}` |



```python
molecule = "Caffeine"                  # str
atom_count = 24                         # int
molecular_weight = 194.19               # float
is_water_soluble = True                 # bool
descriptors = ["LogP", "TPSA", "HBD"]   # list
coordinates = (1.2, 3.4, 5.6)           # tuple
molecule_info = {"name": "Caffeine", "formula": "C8H10N4O2"}  # dict
unique_atoms = {"C", "H", "N", "O"}    # set
```

```python
print("Molecule name is:", molecule)
print("Number of atoms:", atom_count)
print("Molecular weight:", molecular_weight)
print("Is water soluble?", is_water_soluble)
print("Descriptors:", descriptors)
print("Coordinates:", coordinates)
print("Molecule info:", molecule_info)
print("Unique atoms:", unique_atoms)
```

```python
print("molecule dtype is:", type(molecule))
print("atom_count dtype is:", type(atom_count))
print("molecular_weight dtype is:", type(molecular_weight))
print("is_water_soluble dtype is:", type(is_water_soluble))
print("descriptors dtype is:", type(descriptors))
print("coordinates dtype is:", type(coordinates))
print("molecule_info dtype is:", type(molecule_info))
print("unique_atoms dtype is:", type(unique_atoms))
```

```python
# Verifying data types with isinstance()
print(isinstance(molecule, str))
print(isinstance(atom_count, int))
print(isinstance(molecular_weight, float))
print(isinstance(is_water_soluble, bool))
print(isinstance(descriptors, list))
print(isinstance(coordinates, tuple))
print(isinstance(molecule_info, dict))
print(isinstance(unique_atoms, set))
```

## 5. Keywords and Identifiers

### Keywords
**Keywords** are reserved words that have special meanings in Python. Examples include `if`, `else`, `for`, `while`, `True`, `False`, and `None`.

You cannot use a keyword as a variable name.

You can inspect Python's keyword list using the `keyword` module.

## 8. Python Operators

Operators perform calculations, compare values, and combine conditions.

### 8.1 Arithmetic operators

| Operator | Operation | Example |
|---|---|---|
| `+` | Addition | `10 + 3` |
| `-` | Subtraction | `10 - 3` |
| `*` | Multiplication | `10 * 3` |
| `/` | Division | `10 / 3` |
| `//` | Floor division | `10 // 3` |
| `%` | Modulus (remainder) | `10 % 3` |
| `**` | Exponentiation | `2 ** 3` |



```python
a = 10
b = 3

print("Summation:", a + b)
print("Subtraction:", a - b)
print("Multiplication:", a * b)
print("Division:", a / b)
print("Floor Division:", a // b)
print("Modulus:", a % b)
print("Exponentiation:", a ** b)
```

```python
# Practical example: Calculating Molecular Weight of Ethanol (C2H6O)
carbon_wt = 12.011 * 2
hydrogen_wt = 1.008 * 6
oxygen_wt = 15.999 * 1

mw_ethanol = carbon_wt + hydrogen_wt + oxygen_wt
print("Calculated MW of Ethanol (C2H6O):", round(mw_ethanol, 3), "g/mol")
```

### 8.2 Comparison operators

Comparison operators compare two values and return either `True` or `False`.

```python
x = 5
y = 10

print("Equal to:", x == y)
print("Not equal to:", x != y)
print("Greater than:", x > y)
print("Less than:", x < y)
print("Greater than or equal to:", x >= y)
print("Less than or equal to:", x <= y)
```

### 8.3 Logical operators

Logical operators combine or reverse conditions:
- `and`: True when both conditions are True.
- `or`: True when at least one condition is True.
- `not`: Reverses a Boolean result.

```python
score = int(input("Enter your score: "))

is_score = True

print(score > 50 and is_score)
print(score > 90 or is_score)
print(not is_score)
```

### 8.4 Assignment operators

Assignment operators assign or update variable values.

```python
molecular_weight = 180.16
print("Initial:", molecular_weight)

molecular_weight += 10
print("After += 10:", molecular_weight)

molecular_weight -= 5
print("After -= 5:", molecular_weight)

molecular_weight *= 2
print("After *= 2:", molecular_weight)

molecular_weight /= 2
print("After /= 2:", molecular_weight)
```

## 9. Conditional Statements: `if`, `elif`, and `else`

Conditional statements allow a program to make decisions.

- `if` checks the first condition.
- `elif` checks another condition if earlier conditions were False.
- `else` runs when none of the preceding conditions were True.

Python uses **indentation** (commonly four spaces) to define the code inside each block.



```python
molecular_weight = float(input("Enter the molecular weight: "))

if molecular_weight < 0:
    print("Molecular weight cannot be negative.")
elif molecular_weight > 500:
    print("The molecule is heavy (exceeds Lipinski's MW rule of 500 Da).")
elif molecular_weight < 100:
    print("The molecule is light (< 100 Da).")
else:
    print("The molecule is within standard drug-like range (100 - 500 Da).")
```

```python
# Checking Lipinski's Rule of Five criteria
molecular_weight = 350.4
log_p = 2.8
h_bond_donors = 3
h_bond_acceptors = 6

if molecular_weight <= 500 and log_p <= 5 and h_bond_donors <= 5 and h_bond_acceptors <= 10:
    print("Drug Candidate: PASSES Lipinski's Rule of Five.")
else:
    print("Drug Candidate: FAILS Lipinski's Rule of Five.")
```

## 10. `for` Loops

A **for loop** repeats a block of code for each item in a sequence or iterable, such as a list or a range of numbers.


- **range(stop)** - Generates numbers from 0 up to (but not including) stop.

- **range(start, stop)** - Generates numbers from start up to (but not including) stop.

- **range(start, stop, step)** - Controls the increment or decrement between numbers.

- **Counting backward**

- **Direct Iteration** - Iterating directly over lists, tuples, strings, and sets.

- **enumerate()** - Access both index and value during iteration.

- **zip()** - Iterate over two or more collections simultaneously.

**range(stop)**

```python
for i in range(10):
  print(i)
```

**range(start, stop)**

```python
for i in range(1, 10):
  print(i)
```

**range(start, stop, step)**

```python
for i in range(1, 10, 4):
  print(i)
```

```python
# Counting backward using range with a negative step
for i in range(10, 0, -2):
    print(i)
```

**Iterating over a List**

```python
molecule_names = ["Aspirin", "Caffeine", "Paracetamol"]

# Direct iteration over a list
for molecule in molecule_names:
    print("Drug molecule:", molecule)
```

**Iterating over a String**

```python
sequence = "ATGCGA"

# Iterating character by character over a string
for base in sequence:
    print("Nucleotide:", base)
```

**Iterating over a Tuple**

```python
coordinates = (1.2, 3.4, 5.6)

# Iterating over a tuple
for coord in coordinates:
    print("Coordinate value:", coord)
```

**Iterating over a Set**

```python
unique_elements = {"C", "H", "O", "N", "S"}

# Iterating over a set
for element in unique_elements:
    print("Element:", element)
```

**enumerate()**

```python
compounds = ["Aspirin", "Ibuprofen", "Paracetamol"]

# Using enumerate() to get index and value
for index, compound in enumerate(compounds, start=1):
    print(f"Rank {index}: {compound}")
```

```python
# Multiplication table
for i in range(1, 11):
    print(f"5 * {i} = {5 * i}")

print()

# Iterating over multiple collections simultaneously with zip()
drugs = ["Aspirin", "Caffeine", "Paracetamol"]
mw_list = [180.16, 194.19, 151.16]

for drug, mw in zip(drugs, mw_list):
    print(f"Compound: {drug:<12} | MW: {mw} g/mol")
```

## 11. `while` Loops

A **while loop** repeats a block of code as long as its condition remains True.

```python
count = 1

while count < 10:
  print(count)
  count += 3
```

### `for` loop vs `while` loop

| `for` loop | `while` loop |
|---|---|
| Iterates over a sequence or iterable | Repeats while a condition is True |
| Useful when iterating through items or a known range | Useful when the number of repetitions depends on a condition |

## 12. Nested Loops

A **nested loop** is a loop inside another loop. For every iteration of the outer loop, the inner loop completes all its iterations.

Nested loops are useful for working with grids, tables, matrices, and combinations.

```python
for row in range(1, 5):
  for column in range(1, 5):
    print("row:", row, "| column:", column)
  print()
```

```python
# Nested loops: Screening candidate ligands across biological targets
targets = ["EGFR", "CDK2", "BRAF"]
ligands = ["Ligand_1", "Ligand_2"]

for target in targets:
    for ligand in ligands:
        print(f"Docking {ligand} against target {target}")
    print("-" * 35)
```
